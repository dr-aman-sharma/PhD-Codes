```python
def dmrg_prod(W,M1,M2):
    
        #W=MPOs,M=MPSs
        #Iteratively performing the products at the indvidual sites of MPOs and MPSs and
        # multiplying the new tensor with the existing chain of tensors
        #n is the site number which starts from zero and goes upto N-1.
        #N-1 is the last site number
        #N is the number of sites
        
        N=len(M1)
        L=len(M1)
            
        #First complete the left side
        prod_L=1
        prod_R=1
        
        for i in range(N):
        
            if i==0:
            
                prod1=np.tensordot(M1[i],W[i],axes=([0],[0]))#first MPS and MPO multiplied together
                prod2=np.tensordot(prod1,M2[i].conj(),axes=([2],[0]))
                prod_L=prod2
                
                prod1=np.tensordot(prod_L,M1[i+1],axes=([1],[1]))#first MPS and MPO multiplied together
                prod2=np.tensordot(prod1,W[i+1],axes=([1,4],[2,0]))
                prod_L=np.tensordot(prod2,M2[i+1].conj(),axes=([2,4],[1,0]))
                
                
                
            if i>1 and i<N-1:
            
                prod1=np.tensordot(prod_L,M1[i],axes=([2],[1]))#first MPS and MPO multiplied together
                prod2=np.tensordot(prod1,W[i],axes=([2,4],[2,0]))
        
                prod_L=np.tensordot(prod2,M2[i].conj(),axes=([2,4],[1,0]))
                
    
            
            if i==N-1:
            
                prod1=np.tensordot(prod_L,M1[i],axes=([0,2],[2,1]))#first MPS and MPO multiplied together
                prod2=np.tensordot(prod1,W[i],axes=([1,3],[2,0]))
            
            
              
                prod_L=np.tensordot(prod2,M2[i].conj(),axes=([0,1,2],[2,1,0]))        
    

              
    
                
        return prod_L

    
    
    
def dmrg_contract2(M1,M2):
    
        N=len(M1)
    
            
        #First complete the left side
        prod_L=1
        
        
        for i in range(N):
        
            if i==0:
            
                prod_L=np.tensordot(M1[i].conj(),M2[i],axes=([0],[0]))#first MPS and MPO multiplied together
                
                prod_L=np.tensordot(prod_L,M2[i+1].conj(),axes=([1],[1]))
                prod_L=np.tensordot(prod_L,M2[i+1],axes=([2,3],[1,0]))
                
            
                
            if i>1 and i<N-1:
            
                prod_L=np.tensordot(prod_L,M1[i].conj(),axes=([2],[1]))#first MPS and MPO multiplied together
            
        
                prod_L=np.tensordot(prod_L,M2[i],axes=([2,3],[1,0]))
                
               
            
            if i==N-1:
            
                prod_L=np.tensordot(prod_L,M1[i].conj(),axes=([0,2],[2,1]))#first MPS and MPO multiplied together
            
            
            
              
                prod_L=np.tensordot(prod_L,M2[i],axes=([0,1,2],[2,1,0]))        
    

              
               
        return prod_L
```

```python
def dmrg_contract2(M1,M2):
    
        N=len(M1)
    
            
        #First complete the left side
        prod_L=1
        
        
        for i in range(N):
        
            if i==0:
            
                prod_L=np.tensordot(M1[i].conj(),M2[i],axes=([0],[0]))#first MPS and MPO multiplied together
               
                
            
                
            if i>0 and i<N-1:
            
                prod_L=np.tensordot(prod_L,M1[i].conj(),axes=([0],[1]))#first MPS and MPO multiplied together
            
        
                prod_L=np.tensordot(prod_L,M2[i],axes=([0,1],[1,0]))
                
               
            
            if i==N-1:
            
                prod_L=np.tensordot(prod_L,M1[i].conj(),axes=([0],[1]))#first MPS and MPO multiplied together
            
            
            
              
                prod_L=np.tensordot(prod_L,M2[i],axes=([0,1],[1,0]))        
    

              
               
        return prod_L

    
def dmrg_prod(W,M1,M2):
    
        #W=MPOs,M=MPSs
        #Iteratively performing the products at the indvidual sites of MPOs and MPSs and
        # multiplying the new tensor with the existing chain of tensors
        #n is the site number which starts from zero and goes upto N-1.
        #N-1 is the last site number
        #N is the number of sites
        
        N=len(M1)
        L=len(M1)
            
        #First complete the left side
        prod_L=1
        prod_R=1
        
        for i in range(N):
        
            if i==0:
            
                prod1=np.tensordot(M1[i],W[i],axes=([0],[0]))#first MPS and MPO multiplied together
                prod2=np.tensordot(prod1,M2[i].conj(),axes=([1],[0]))
                prod_L=prod2
                
            if i>0 and i<N-1:
            
                prod1=np.tensordot(prod_L,M1[i],axes=([0],[1]))#first MPS and MPO multiplied together
                prod2=np.tensordot(prod1,W[i],axes=([0,2],[2,0]))
        
                prod_L=np.tensordot(prod2,M2[i].conj(),axes=([0,2],[1,0]))
                
    
            
            if i==N-1:
            
                prod1=np.tensordot(prod_L,M1[i],axes=([0],[1]))#first MPS and MPO multiplied together
                prod2=np.tensordot(prod1,W[i],axes=([0,2],[2,0]))
            
            
              
                prod_L=np.tensordot(prod2,M2[i].conj(),axes=([0,1],[1,0]))        
    

              
    
                
        return prod_L
```

```python
#new spinon


def mpo1(M1,M2,l):
    
    for i in range(l):
            if i==0:
            
                prod1=np.tensordot(M1[i],HF,axes=([0],[0]))#first MPS and MPO multiplied together
                prod2=np.tensordot(prod1,M2[i].conj(),axes=([2],[0]))
                prod_L=prod2
                
                prod1=np.tensordot(prod_L,M1[i+1],axes=([1],[1]))#first MPS and MPO multiplied together
                prod2=np.tensordot(prod1,H,axes=([1,4],[2,0]))
                prod_L=np.tensordot(prod2,M2[i+1].conj(),axes=([2,4],[1,0]))
                
                
                
            if i>1:
            
                prod1=np.tensordot(prod_L,M1[i],axes=([2],[1]))#first MPS and MPO multiplied together
                prod2=np.tensordot(prod1,H,axes=([2,4],[2,0]))
        
                prod_L=np.tensordot(prod2,M2[i].conj(),axes=([2,4],[1,0]))
            
            
           
    return(prod_L)

def mpo1a(M1,M2,l):
    
    for i in range(l):
            if i==0:
            
                prod1=np.tensordot(M1[i],H,axes=([0],[0]))#first MPS and MPO multiplied together
                prod2=np.tensordot(prod1,M2[i].conj(),axes=([2],[0]))
                prod_L=prod2
                
                prod1=np.tensordot(prod_L,M1[i+1],axes=([1],[1]))#first MPS and MPO multiplied together
                prod2=np.tensordot(prod1,H,axes=([2,5],[2,0]))
                prod_L=np.tensordot(prod2,M2[i+1].conj(),axes=([3,5],[1,0]))
                
                
                
            if i>1:
            
                prod1=np.tensordot(prod_L,M1[i],axes=([3],[1]))#first MPS and MPO multiplied together
                prod2=np.tensordot(prod1,H,axes=([3,5],[2,0]))
        
                prod_L=np.tensordot(prod2,M2[i].conj(),axes=([3,5],[1,0]))
            
            
           
    return(prod_L)
        
        
def mpo2(M1,M2,l):
    
    if l==1:
        i=0
        prod1=np.tensordot(M1[i],H,axes=([0],[0]))#first MPS and MPO multiplied together
        prod2=np.tensordot(prod1,M2[i].conj(),axes=([3],[0]))
        prod_L=prod2
        
       
    
    if l>1:
        
        for i in range(l):
        
            if i==0:
                
                prod1=np.tensordot(M1[i],H,axes=([0],[0]))#first MPS and MPO multiplied together
                prod2=np.tensordot(prod1,M2[i].conj(),axes=([3],[0]))
                prod_L=prod2
                
            if i==1:
                
                prod1=np.tensordot(prod_L,M1[i],axes=([1],[1]))#first MPS and MPO multiplied together
                prod2=np.tensordot(prod1,H,axes=([3,6],[2,0]))
                prod_L=np.tensordot(prod2,M2[i].conj(),axes=([4,6],[1,0]))
                
            if i>1:
                
                
                prod1=np.tensordot(prod_L,M1[i],axes=([4],[1]))#first MPS and MPO multiplied together
                prod2=np.tensordot(prod1,H,axes=([4,6],[2,0]))
                prod_L=np.tensordot(prod2,M2[i].conj(),axes=([4,6],[1,0]))
                
    return(prod_L)

def mpo3(M1,M2,l):
    
    
    prod_L=mpo1a(M1,M2,l-1)
    
    
    
    prod1=np.tensordot(prod_L,spinon,axes=([3],[1]))
    prod2=np.tensordot(prod1,HL,axes=([3,5],[2,0]))
    prod_L=np.tensordot(prod2,spinon,axes=([3,6],[1,0]))
    
    return(prod_L)
        
        
def grandmpo(M1,M2,j):
    
    l1=len(M1)//2
    l2=j-len(M1)//2+1
    l3=len(M1)-j-1
     
    mpoa=mpo1(M1,M2,l1)
    mpob=mpo2(M1[l1:],M2[l1:],l2)
    mpoc=mpo3(M1[l1+l2:],M2[l1+l2:],l3)
    


    
    
    if l2==1:
        
        prod1=np.tensordot(mpoa,mpob,axes=([2,3,4],[0,3,5]))
        prod_L=np.tensordot(prod1,mpoc,axes=([2,4,5,0,1,3,6],[0,1,2,3,5,4,6]))
        
    if l2>1:
        
        prod1=np.tensordot(mpoa,mpob,axes=([2,3,4],[0,2,3]))
        prod_L=np.tensordot(prod1,mpoc,axes=([3,4,5,0,1,2,6],[0,1,2,3,5,4,6]))
        
        
        
    return(prod_L)

def mps1(M1,M2,l):
    
    for i in range(l):
        
            if i==0:
            
                prod_L=np.tensordot(M1[i].conj(),M2[i],axes=([0],[0]))#first MPS and MPO multiplied together
                
                prod_L=np.tensordot(prod_L,M1[i+1].conj(),axes=([1],[1]))
                prod_L=np.tensordot(prod_L,M2[i+1],axes=([2,3],[1,0]))
                
            
                
            if i>1:
            
                prod_L=np.tensordot(prod_L,M1[i].conj(),axes=([2],[1]))#first MPS and MPO multiplied together
            
                prod_L=np.tensordot(prod_L,M2[i],axes=([2,3],[1,0]))
                
    return(prod_L)
    
    
def mps2(M1,M2,l):
    
    
    if l==1:
        
        i=0
        
        prod_L=np.tensordot(M1[i],M2[i],axes=([0],[0]))#first MPS and MPO multiplied together
       
    
    if l>1:
        for i in range(l):
        
            if i==0:
                
                prod1=np.tensordot(M1[i],M2[i],axes=([0],[0]))#first MPS and MPO multiplied together
                prod_L=prod1
                
            if i==1:
                
                prod1=np.tensordot(prod_L,M1[i],axes=([1],[1]))#first MPS and MPO multiplied together
                prod_L=np.tensordot(prod1,M2[i].conj(),axes=([3,4],[1,0]))
                
            if i>1:
                
                
                prod1=np.tensordot(prod_L,M1[i],axes=([3],[1]))#first MPS and MPO multiplied together
                prod_L=np.tensordot(prod1,M2[i].conj(),axes=([3,4],[1,0]))
                
    return(prod_L)
    
    
    
    
    
    
def mps3(M1,M2,l):
    
    prod_L=mps1(M1,M2,l-1)
    
    
    prod1=np.tensordot(prod_L,spinon,axes=([2],[1]))
    prod_L=np.tensordot(prod1,spinon,axes=([2,3],[1,0]))
    
    return(prod_L)
    

def grandmps(M1,M2,j):
    
    l1=len(M1)//2
    l2=j-len(M1)//2+1
    l3=len(M1)-j-1
    
    mpsa=mps1(M1,M2,l1)
    mpsb=mps2(M1[l1:],M2[l1:],l2)
    mpsc=mps3(M1[l1+l2:],M2[l1+l2:],l3)
    
    if l2==1:
        
        prod1=np.tensordot(mpsa,mpsb,axes=([2,3],[0,3]))
        prod_L=np.tensordot(prod1,mpsc,axes=([2,4,0,1,3,5],[0,1,2,4,3,5]))
        

    if l2>1:
        
        prod1=np.tensordot(mpsa,mpsb,axes=([2,3],[0,2]))
        prod_L=np.tensordot(prod1,mpsc,axes=([3,4,0,1,2,5],[0,1,2,4,3,5]))
        
    return(prod_L)




    
```
