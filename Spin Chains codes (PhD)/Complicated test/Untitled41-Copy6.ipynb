```python
import matplotlib.pyplot as plt
from matplotlib import colors
import pickle
import numpy as np

f=open("L=80,BD=160,30s","rb")
spacetime,E,tau,M,Num=pickle.load(f)

F=

spacetime=spacetime[len(spacetime)//2-int(F/tau):len(spacetime)//2+int(F/tau)+1,:]
S=0
E=F
Num=(E-S)/tau+1




#Fourier Transform

k=np.linspace(0,6.28,100)
og=np.linspace(4,0,100)

k,og=np.meshgrid(k,og)



#Space Column

sigm_t=E/4
sigm_s=(len(M)-1)/4

time=0
index=0
for j in np.linspace(-E,E,2*int(Num)-1):
    
    space=0
    
    for i in range(len(M)):
        space=space+np.exp(1j*k*(len(M)//2-i))*spacetime[index,i]*np.exp(-(len(M)//2-i)**2/sigm_s**2)
    
    print(j)
    time=time+np.exp(1j*og*j)*space*tau*np.exp(-j**2/sigm_t**2)#*(1/np.sqrt(np.pi*sigm_t**2))
    
    
    index=index+1
    
#time=time*time.conj()
print(time)
    
ax=plt.axes(projection='3d')

ax.plot_surface(k,og,time)

fig, ax = plt.subplots()
extent=0,6.28,0,4
im = ax.imshow(np.abs(time),extent=extent,interpolation='gaussian',cmap="jet",norm=colors.LogNorm(vmin=0.4))
fig.colorbar(im)

plt.show()
```
