```python
import numpy as np
import matplotlib.pyplot as plt

# Define the parameter J2
J2 = 0.43 # You can set J2 to any constant value you need

# Define the single magnon dispersion function Delta E_q
def delta_E_q(q, J2):
    numerator = (5/3) * ((5/2) - J2 * (1 - np.cos(2*q))) * (37 - 12 * np.cos(2*q))
    denominator = 101 - 24 * np.cos(2*q)
    return numerator / denominator

# Generate a range of q values
q_values = np.linspace(0, np.pi, 500)

# Calculate Delta E_q for each q value
delta_E_q_values = delta_E_q(q_values, J2)

# Initialize lists to store q and E values for the 2-magnon continuum
q_continuum = []
E_continuum = []

# Calculate the 2-magnon continuum
for q1 in q_values:
    for q2 in q_values:
        q_sum = q1 + q2
        if q_sum <= 2*np.pi:  # We consider q_sum in the range [0, 2*pi]
            E_total = delta_E_q(q1, J2) + delta_E_q(q2, J2)
            q_continuum.append(q_sum)
            E_continuum.append(E_total)

# Convert lists to numpy arrays for plotting
q_continuum = np.array(q_continuum)
E_continuum = np.array(E_continuum)

# Plot the 2-magnon continuum
plt.figure(figsize=(10, 6))
plt.scatter(q_continuum, E_continuum, s=1, color='blue')
plt.xlabel(r'$q = q_1 + q_2$')
plt.ylabel(r'$E(q)$')
plt.title('2-Magnon Continuum')
plt.grid(True)
plt.ylim([0,10])
plt.xlim([0,2*np.pi])
plt.show()
```

```python
def find_index_of_smallest_element(arr):
    # Initialize with the first index
    smallest_index = 0

    # Loop through the array starting from the second element
    for i in range(1, len(arr)):
        if arr[i] < arr[smallest_index]:
            smallest_index = i

    return smallest_index

# Example usage:

index = find_index_of_smallest_element(Og2)
```
